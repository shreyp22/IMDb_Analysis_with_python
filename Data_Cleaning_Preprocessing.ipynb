## IV. Release Date

<span style='font-size:18px;'>
➤ A small portion of records had missing release dates due to incomplete JSON data. However, month and year information was available on the webpage. <br><br>
    
➤ Missing entries were re-scraped from the source <br>

➤ The first day of the respective month was assigned **(i.e., day=1)** <br>

➤ Default date was assigned as **‘2000-01-01’** for null records even after re-scraping. We got 7 records out of 5649, so replaced by default.
</span>

```python
df['release_date'].isna().sum()
```

```python
missing_df = df[df['release_date'].isna()]
missing_df
```

```python
from datetime import datetime

# converting to date from the month and year
def create_date(month, year):
    try:
        return datetime.strptime(f"1 {month} {year}", "%d %B %Y").date()
    except:
        return None
```

```python
import re, time, random
from bs4 import BeautifulSoup

# Getting month and year from the movie page
def scrape_month_year(driver, url):
    driver.get(url)
    
    time.sleep(random.uniform(3, 5))
    
    soup = BeautifulSoup(driver.page_source, "html.parser")

    try:
        block = soup.find("div", {"data-testid": "title-details-section"})
        date_text = block.select_one('[data-testid="title-details-releasedate"] div').text.strip()
    except:
        date_text = '2000-01-01'

    match = re.match(r'(.*\d).*', date_text)

    try:
        month, year = match.group(1).split(' ')
    except:
        year = match.group(1)
        month = 'January'
        
    return month, year
```

```python
from IMDB_Scraping import get_driver

driver = get_driver()
count = 1

# Re-scraping and replacing null dates
for idx, row in missing_df.iterrows():
    url = row['movie_url']
    print(f"Re-Scraping: {count}/{len(missing_df)}")
    month, year = scrape_month_year(driver, url)
    df.loc[idx, "release_date"] = create_date(month, year)

    time.sleep(random.uniform(2,4))
    count += 1

driver.quit()
```

# VII. Watchlist Count, User Reviews, & Critic Reviews

<span style='font-size:18px;'>
➤ A small portion of records had missing values in watchlist_count, user_reviews, and critic_reviews due to incomplete information about it on the movie page. <br>

➤ Missing entries were replaced with **'0'**, because as it was not present which means no one had given the reviews nor added to the watchlist. <br>
➤ Values in **'K'** and **'M'** are converted into numeric form (1000s and 1000000s).
</span>

```python
def convert_to_number(x):
    if pd.isna(x):
        return 0

    if isinstance(x, (int, float)):
        return x
    
    x = str(x).replace(',', '').strip().upper()
    
    if 'K' in x:
        return float(x.replace('K', '')) * 1_000
    elif 'M' in x:
        return float(x.replace('M', '')) * 1_000_000
    else:
        return float(x)
```

```python
df['watchlist_count'] = df['watchlist_count'].apply(convert_to_number)
df['user_reviews'] = df['user_reviews'].apply(convert_to_number)
df['critic_reviews'] = df['critic_reviews'].apply(convert_to_number)
```

# Currency Conversion

```python
df['budget'] = df['budget'].astype(str)
df['gross'] = df['gross'].astype(str)
```

```python
import re

def extract_currency_type(x):
    if pd.isnull(x):
        return None
    
    # Extract currency prefix (before numbers)
    match = re.match(r"[^\d]+", x)
    return match.group().strip() if match else None


df["currency_type"] = df["budget"].apply(extract_currency_type)
```

```python
df['currency_type'].unique()
```

```python
conversion_rates = {
    "$": 1.0,        # USD
    "£": 1.25,       # GBP
    "€": 1.10,       # Euro
    "¥": 0.007,      # Japanese Yen
    "CN¥": 0.14,     # Chinese Yuan
    "₹": 0.012,      # Indian Rupee
    "A$": 0.65,      # Australian Dollar
    "CA$": 0.75,     # Canadian Dollar
    "NZ$": 0.60,     # New Zealand Dollar
    "R$": 0.20,      # Brazilian Real
    "MX$": 0.058,    # Mexican Peso
    "NT$": 0.031,    # Taiwan Dollar
    "₩": 0.00075,    # South Korean Won
    "SEK": 0.095,    # Swedish Krona
    "NOK": 0.095,    # Norwegian Krone
    "DKK": 0.15,     # Danish Krone
    "PLN": 0.25,     # Polish Zloty
    "HUF": 0.0028,   # Hungarian Forint
    "RON": 0.22,     # Romanian Leu
    "ZAR": 0.055,    # South African Rand
    "SAR": 0.27,     # Saudi Riyal
    "EGP": 0.032,    # Egyptian Pound
    "PKR": 0.0036,   # Pakistani Rupee
    "BDT": 0.0091,   # Bangladeshi Taka
    "NPR": 0.0075,   # Nepali Rupee
    "LKR": 0.0031,   # Sri Lankan Rupee
    "IDR": 0.000065, # Indonesian Rupiah
    "MVR": 0.065,    # Maldivian Rufiyaa

    # Old currencies (approx historical conversion)
    "ITL": 0.00055,  # Italian Lira
    "FRF": 0.18,     # French Franc
    "DEM": 0.55,     # German Mark
    "TRL": 0.000000035, # Old Turkish Lira
    "ATS": 0.073,    # Austrian Schilling
    "RUR": 0.013,    # Old Russian Ruble

    # Missing / unknown
    "nan": None
}
```

```python
def extract_currency_amount(x):
    if pd.isnull(x):
        return None, None
    
    currency = re.match(r"[^\d]+", x)
    currency = currency.group().strip() if currency else None
    
    amount = re.findall(r"[\d,]+", x)
    amount = int(amount[0].replace(",", "")) if amount else None
    
    return currency, amount
```

```python
def convert_to_usd(x):
    currency, amount = extract_currency_amount(x)
    
    if currency in conversion_rates and amount:
        rate = conversion_rates[currency]
        if rate:
            return amount * rate
    
    return None
```

```python
df['gross_usd'] = df['gross'].apply(convert_to_usd)
df["budget_usd"] = df["budget"].apply(convert_to_usd)
```
